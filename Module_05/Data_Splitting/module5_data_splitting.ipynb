```python
##############################################################
# Import the libraries that I will need for this exercise
##############################################################

from sklearn.datasets import load_wine
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, classification_report


##############################################################
# Load the Wine dataset
# Here I separate the input variables (X) from the labels (y)
##############################################################

data = load_wine()

X = data.data
y = data.target


##############################################################
# Split the data
#
# The aim is to obtain:
# - 60% training data
# - 20% validation data
# - 20% test data
#
# As train_test_split() only creates two groups,
# I need to do the split in two different steps.
##############################################################

# First I keep 20% of the data as the final test set.
# The remaining 80% will be used for training and validation.
X_train_val, X_test, y_train_val, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

# Now I split the remaining 80%.
# I take 25% of it for validation.
# This corresponds to 20% of the original dataset.
X_train, X_val, y_train, y_val = train_test_split(
    X_train_val,
    y_train_val,
    test_size=0.25,
    random_state=42,
    stratify=y_train_val
)


##############################################################
# Create and train the Logistic Regression model
##############################################################

model = LogisticRegression(
    max_iter=5000,
    random_state=42
)

# Train the model using only the training data
model.fit(X_train, y_train)


##############################################################
# Evaluate the model using the validation set
#
# I first make predictions and then calculate
# the validation accuracy.
##############################################################

val_preds = model.predict(X_val)

val_accuracy = accuracy_score(y_val, val_preds)

print(f"Validation Accuracy: {val_accuracy:.4f}")


##############################################################
# Evaluate the model on the test set
#
# This gives an estimate of how well the model
# performs on unseen data.
##############################################################

test_preds = model.predict(X_test)

test_accuracy = accuracy_score(y_test, test_preds)

print(f"Test Accuracy: {test_accuracy:.4f}")


##############################################################
# Print a more detailed summary of the results
#
# This report includes precision, recall,
# F1-score and support for each class.
##############################################################

print("\nClassification Report on Test Set:")
print(
    classification_report(
        y_test,
        test_preds,
        target_names=data.target_names
    )
)
```
